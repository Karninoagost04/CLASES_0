```python
# Importar librerías
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, clear_output


from tensorflow.keras.datasets import mnist

(X_train, y_train), (X_test, y_test) = mnist.load_data()
```

```python
print(X_train.shape)
print(y_train.shape)
print(X_test.shape)
print(y_test.shape)
```

```python
# Campo para seleccionar la imagen
indice = widgets.IntText(
    value=0,
    description='Imagen:',
    min=0,
    max=len(X_train)-1
)

boton = widgets.Button(
    description='Mostrar imagen'
)

salida = widgets.Output()
def mostrar_imagen(b):
    with salida:
        clear_output(wait=True)

        i = indice.value

        plt.figure(figsize=(4,4))
        plt.imshow(X_train[i], cmap='gray')
        plt.title(f'Imagen {i} — Dígito: {y_train[i]}')
        plt.axis('off')
        plt.show()

        print("Tamaño de la imagen:", X_train[i].shape)
        print("Etiqueta:", y_train[i])

boton.on_click(mostrar_imagen)

display(indice, boton, salida)
```
