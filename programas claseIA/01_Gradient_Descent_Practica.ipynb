J(θ)=(θ−3)2

```python
# Parámetro que queremos encontrar
theta = 0

# Función de costo
def J(theta):
    return (theta - 3)**2

print("Theta:", theta)
print("Error:", J(theta))
```

```python
# Calculamos el gradiente

def gradiente(theta):
    return 2 * (theta - 3)

print("Theta:", theta)
print("Gradiente:", gradiente(theta))
```

```python
# Learning rate
alpha = 0.1

# Calculamos el nuevo theta
theta_nuevo = theta - alpha * gradiente(theta)

print("Theta anterior:", theta)
print("Gradiente:", gradiente(theta))
print("Learning rate:", alpha)
print("Theta nuevo:", theta_nuevo)
```
